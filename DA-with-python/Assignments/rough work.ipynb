```python
num = [-5, -3, 1, 2, 100, -99, 0,55]
positive = []
negative = []
Zero = []

for n in num:
    if n < 0:
        print(f"{n} is negative")
        negative.append(n)
    elif n == 0:
        print(f"{n} is zero")
        Zero.append(n)
    else:
        print(f"{n} is positive")
        positive.append(n)
print(f"Positive numbers: {positive}")
print(f"Negative numbers: {negative}")
print(f"Zero numbers: {Zero}")
```

```python
n = int(input("Enter a number: "))
factorial = 1
count = 1
while count <= n:
    factorial = factorial * count
    count = count + 1
print(factorial)

```

```python
def grade_system(score):
    if score > 90:
        return "A+"
    elif score >= 80 and score <= 90:
        return "A"
    elif score >= 70 and score < 80:
        return "B"
    elif score >= 60 and score < 70:
        return "C"
    else:
        return "F"


def student_grade(name, score):
    grade = grade_system(score)
    print(f"{name} scored {score} and received a grade of {grade}.")


student_grade("Alice", 85)
```
